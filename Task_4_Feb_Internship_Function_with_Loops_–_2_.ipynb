Problem 1: Mobile Recharge Validation System


```python
def validate_recharge():
    valid_plans = [199, 299, 399, 599]

    while True:
        amount = int(input("Enter recharge amount: "))

        if amount < 50:
            print("Recharge amount must be at least ₹50.")
        elif amount not in valid_plans:
            print("Invalid plan selected.")
        else:
            print("Recharge successful!")
            break
```

Problem 2: Inventory Reorder Alert System


```python
def check_inventory(products):
    for item, quantity in products.items():
        if quantity < 15:
            print(f"{item}: Reorder Alert")
        else:
            print(f"{item}: Stock OK")

inventory = {
    "Pants": 10,
    "Shirts": 25,
    "Bags": 12
}

check_inventory(inventory)
```

Problem 3: Student Result Processing System

```python
def process_result(marks):
    total = 0

    for mark in marks:
        total += mark

    average = total / len(marks)

    if average >= 50:
        return "Pass"
    else:
        return "Fail"

marks_list = [66, 77, 44,55, 88]
result = process_result(marks_list)

print("Result:", result)
```

Problem 4: Cab Fare Estimator with Retry Option


```python
def calculate_fare(distance, peak):
    fare = 50 + (12 * distance)

    if peak.lower() == "yes":
        fare *= 1.25

    return fare

while True:
    km = float(input("Enter distance in km: "))
    peak_hour = input("Is it peak hour? (yes/no): ")

    total_fare = calculate_fare(km, peak_hour)
    print("Total Fare: ₹", total_fare)

    retry = input("Calculate again? (yes/no): ")
    if retry.lower() != "yes":
        break
```

Problem 5: Employee Attendance Eligibility

```python
def check_eligibility(attendance):
    present_days = 0

    for day in attendance:
        if day == "P":
            present_days += 1

    percentage = (present_days / len(attendance)) * 100

    if percentage >= 75:
        return "Eligible"
    else:
        return "Not Eligible"

attendance_list = ["P", "P", "A", "P", "P"]
print(check_eligibility(attendance_list))
```

Problem 6: Password Strength Checker


```python
def check_password(password):
    has_digit = False
    has_special = False

    for char in password:
        if char.isdigit():
            has_digit = True
        if char in "@#$":
            has_special = True

    if len(password) >= 8 and has_digit and has_special:
        return "Strong Password"
    else:
        return "Weak Password"

user_password = input("Enter password: ")
print(check_password(user_password))
```
