Problem 1: Social Media – Post Engagement Analyzer



```python
def analyze_post_engagement(likes_list):
    total_likes = 0

    for likes in likes_list:
        total_likes += likes

    if total_likes >= 1000:
        status = "Viral Post"
    else:
        status = "Normal Engagement"

    print("Total Likes:", total_likes)
    print("Post Status:", status)

# Example
analyze_post_engagement([200, 300, 250, 300])
```

Problem 3: Agriculture – Rainfall Adequacy Checker


```python
def rainfall_checker(rainfall_list, required_level):
    total = 0

    for rain in rainfall_list:
        total += rain

    average = total / len(rainfall_list)

    if average >= required_level:
        status = "Adequate Rainfall"
    else:
        status = "Inadequate Rainfall"

    print("Average Rainfall:", average)
    print("Rainfall Status:", status)

# Example
rainfall_checker([70, 75, 72, 71], 70)
```

Problem 4: Social Media – Duplicate Account Detection


```python
def detect_duplicates(usernames):
    if len(usernames) != len(set(usernames)):
        print("Duplicate Accounts Found: Yes")
    else:
        print("Duplicate Accounts Found: No")

# Example
detect_duplicates(["user1", "user2", "user1"])
```

Problem 6: Agriculture – Premium Crop Price Filter


```python
def filter_premium_crops(prices):
    premium = []

    for price in prices:
        if price > 2000:
            premium.append(price)

    print("Premium Crops:", premium)

# Example
filter_premium_crops([1500, 2500, 1800, 3200])
```

Problem 7: System Monitoring – Application Health Checker


```python
def check_system_health(errors):
    if errors == 0:
        status = "Healthy"
    elif errors <= 5:
        status = "Minor Issues"
    else:
        status = "Critical Issues"

    print("Error Count:", errors)
    print("System Status:", status)

# Example
check_system_health(7)
```

Problem 9: E-Learning – Student Attendance Eligibility System


```python
def check_attendance(attendance_list):
    total = 0

    for record in attendance_list:
        total += record

    percentage = (total / len(attendance_list)) * 100

    if percentage >= 75:
        status = "Eligible"
    else:
        status = "Not Eligible"

    print("Attendance Percentage:", percentage)
    print("Exam Eligibility:", status)

# Example (1 = Present, 0 = Absent)
check_attendance([1,1,1,1,0])
```

Problem 10: Smart Electricity Bill Analyzer


```python
def calculate_bill(units):
    bill = 0

    if units <= 100:
        bill = units * 3
    elif units <= 200:
        bill = (100 * 3) + ((units - 100) * 5)
    else:
        bill = (100 * 3) + (100 * 5) + ((units - 200) * 7)

    if bill < 500:
        usage = "Low Usage"
    elif bill <= 1500:
        usage = "Moderate Usage"
    else:
        usage = "High Usage"

    print("Total Bill:", bill)
    print("Usage Classification:", usage)

# Example
calculate_bill(250)
```
