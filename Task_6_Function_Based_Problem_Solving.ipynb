Problem Statement 1:Smart Parking Lot Management System

```python
def smart_parking(capacity, logs):
    parked = 0
    peak = 0

    for log in logs:
        if log == "IN":
            parked += 1
        elif log == "OUT" and parked > 0:
            parked -= 1

        peak = max(peak, parked)

    print("Currently Parked Vehicles:", parked)

    if parked > capacity:
        print("Parking Status: Full (Over Capacity Alert!)")
    else:
        print("Parking Status: Available")

smart_parking(50, ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"])
```


 Problem Statement 2:Online Food Delivery Time Estimator


```python
def estimate_delivery(distance, traffic, weather):
    base_time = distance * 5   # 5 minutes per km

    # Traffic adjustment
    if traffic == "High":
        base_time += 15
    elif traffic == "Medium":
        base_time += 8

    # Weather adjustment
    if weather == "Rainy":
        base_time += 10
    elif weather == "Stormy":
        base_time += 20

    print("Estimated Delivery Time:", base_time, "minutes")

estimate_delivery(8, "High", "Rainy")
```

Problem Statement 3:Movie Theatre Seat Occupancy

```python
def seat_occupancy(total_seats, booked_list):
    booked = len(booked_list)
    occupancy = (booked / total_seats) * 100

    print("Occupancy:", int(occupancy), "%")

    if occupancy == 100:
        print("Show Status: Housefull")
    elif occupancy >= 75:
        print("Show Status: Almost Full")
    else:
        print("Show Status: Seats Available")

seat_occupancy(200, [1]*150)
```

Problem Statement 4:Cloud Server Load Classification System


```python
def server_load(cpu_readings):
    avg = sum(cpu_readings) / len(cpu_readings)

    print("Average CPU Load:", int(avg), "%")

    if avg < 50:
        print("Server Status: Normal")
    elif 50 <= avg <= 80:
        print("Server Status: Warning")
    else:
        print("Server Status: Critical")

server_load([45, 60, 70, 85, 90])
```


 Problem Statement 5:Smart Classroom Resource Usage Monitor


```python
def resource_monitor(usage):
    overused = []

    for resource, hours in usage.items():
        if hours > 8:   # threshold
            overused.append(resource)

    if overused:
        print("Overused Resources:", ", ".join(overused))
        print("Energy Alert: Yes")
    else:
        print("Overused Resources: None")
        print("Energy Alert: No")

resource_monitor({
    "Projector": 6,
    "AC": 9,
    "Lights": 4
})
```

 Problem Statement 6: Online Event Registration Capacity Controller


```python
def event_registration(capacity, registrations):

    if registrations > capacity:
        confirmed = capacity
        waitlisted = registrations - capacity
        status = "Closed"
    else:
        confirmed = registrations
        waitlisted = 0
        status = "Open"

    print("Confirmed Registrations:", confirmed)
    print("Waitlisted Users:", waitlisted)
    print("Registration Status:", status)

event_registration(100, 105)
```
