Task 1: Conceptual Understanding


### 1. Difference between "Love" and "love" in NLP
In NLP, "Love" and "love" are treated as different tokens in case-sensitive systems.
"Love" may appear at the beginning of a sentence or as a proper noun, while "love" represents the general word.
To avoid duplication and inconsistency, text is usually converted to lowercase.

### 2. What happens if stopwords are not removed?
If stopwords are not removed:
- The dataset becomes noisy
- Model training becomes slower
- Important words get less weight
- Accuracy may decrease

---

### 3. When removing stopwords can be harmful?
1. Sentiment Analysis:
   "I am not happy" → removing "not" changes meaning

2. Question Answering:
   "What is your name?" → removing "what" loses intent

---

### 4. Stemming vs Lemmatization
- Stemming: Cuts words to root form (may be incorrect)
  Example: "running" → "run", "better" → "better"

- Lemmatization: Uses vocabulary and grammar
  Example: "better" → "good"

Lemmatization is more accurate but slower.

Task 2: Advanced Preprocessing Function

```python
import re
import string

# Exception words
EXCEPTIONS = {"no", "not"}

def preprocess_text(text):
    try:
        if not isinstance(text, str) or text.strip() == "":
            return []

        # 1. Lowercase
        text = text.lower()

        # 2. Remove URLs & emails
        text = re.sub(r"http\S+|www\S+|@\S+", "", text)

        # 3. Remove numbers
        text = re.sub(r"\d+", "", text)

        # 4. Remove emojis & non-ascii
        text = text.encode("ascii", "ignore").decode()

        # 5. Handle repeated characters (soooo → so)
        text = re.sub(r"(.)\1{2,}", r"\1", text)

        # 6. Remove punctuation
        text = text.translate(str.maketrans('', '', string.punctuation))

        # 7. Remove extra spaces
        text = re.sub(r"\s+", " ", text).strip()

        # 8. Tokenization
        tokens = text.split()

        # 9. Remove short tokens (<=2) except exceptions
        tokens = [word for word in tokens if len(word) > 2 or word in EXCEPTIONS]

        return tokens

    except Exception as e:
        print("Error:", e)
        return []
```

Task 3: Stress Testing

```python
test_sentences = [
    "Get 100% FREE access now!!!",
    "I absolutely looooved this product 😍😍",
    "Worst service ever... 0/10",
    "Call me at 9876543210",
    "This is THE best course!!!",
    "Visit https://openai.com now!",
    "Nooooo this is baaad!!!",
    "OK OK OK I got it",
    "Win $$$ now!!! Limited offer!!!",
    "I am not happy with this"
]

results = []

for text in test_sentences:
    tokens = preprocess_text(text)
    clean_sentence = " ".join(tokens)

    print("\nOriginal:", text)
    print("Tokens:", tokens)
    print("Cleaned Sentence:", clean_sentence)

    results.append(tokens)
```

Task 4: Token Analytics

```python
import numpy as np

for i, tokens in enumerate(results):
    total = len(tokens)
    unique = len(set(tokens))
    avg_len = np.mean([len(t) for t in tokens]) if tokens else 0

    print(f"\nSentence {i+1}")
    print("Total tokens:", total)
    print("Unique tokens:", unique)
    print("Avg token length:", round(avg_len, 2))
```

Task 5: Frequency Analysis

```python
from collections import Counter

all_tokens = [token for sublist in results for token in sublist]

counter = Counter(all_tokens)

print("Top 10 frequent words:", counter.most_common(10))

# Least frequent (count = 1)
least_common = [word for word, count in counter.items() if count == 1][:5]
print("Top 5 least frequent words:", least_common)
```

Task 6: Full Pipeline

```python
def full_pipeline(text_list):
    all_tokens = []
    clean_sentences = []

    for text in text_list:
        tokens = preprocess_text(text)
        sentence = " ".join(tokens)

        all_tokens.extend(tokens)
        clean_sentences.append(sentence)

    return {
        "tokens": all_tokens,
        "clean_sentences": clean_sentences
    }

# Run pipeline
output = full_pipeline(test_sentences)

print(output)
```
